```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")

df = pd.read_csv("/content/Nassau Candy Distributor.csv")

print("Dataset shape:", df.shape)
df.head()
```

```python
print("Column names:")
for col in df.columns:
    print(col)
```

```python
print("DATA TYPES AND NON-NULL COUNTS")
df.info()

print("\nMISSING VALUES")
print(df.isnull().sum())

print("\nTOTAL MISSING VALUES:", df.isnull().sum().sum())
```

```python
order_dates = pd.to_datetime(df["Order Date"], dayfirst=True, errors="coerce")
ship_dates = pd.to_datetime(df["Ship Date"], dayfirst=True, errors="coerce")
lead_time = (ship_dates - order_dates).dt.days

print("Invalid order dates:", order_dates.isna().sum())
print("Invalid ship dates:", ship_dates.isna().sum())
print("Ship dates before order dates:", (lead_time < 0).sum())

print("\nLead-time statistics:")
print(lead_time.describe())
```

```python

# 1. Inspect the original date values
print("Original date values:")
print(df[["Order Date", "Ship Date"]].head(10).to_string(index=False))

# 2. Check the earliest and latest dates
print("\nOrder date range:")
print("Earliest:", order_dates.min())
print("Latest:", order_dates.max())

print("\nShip date range:")
print("Earliest:", ship_dates.min())
print("Latest:", ship_dates.max())

# 3. Check which years appear in the data
print("\nOrder counts by year:")
print(order_dates.dt.year.value_counts().sort_index())

print("\nShip counts by year:")
print(ship_dates.dt.year.value_counts().sort_index())

# 4. Inspect the shortest and longest calculated durations
date_check = df[["Order Date", "Ship Date"]].copy()
date_check["Lead Time (days)"] = lead_time

print("\nShortest lead times:")
print(date_check.nsmallest(5, "Lead Time (days)").to_string(index=False))

print("\nLongest lead times:")
print(date_check.nlargest(5, "Lead Time (days)").to_string(index=False))
```

```python

# Check order ID patterns
print("Sample order IDs:")
print(df["Order ID"].head(15).to_list())

# Check shipping mode counts
print("\nShipping mode counts:")
print(df["Ship Mode"].value_counts())

# Check lead time by shipping mode
date_analysis = df[["Ship Mode"]].copy()
date_analysis["Lead Time (days)"] = lead_time

print("\nLead time by shipping mode:")
print(date_analysis.groupby("Ship Mode")["Lead Time (days)"].agg(["count", "mean", "median", "min", "max"]).round(2))

# Check whether different rows of the same order
# have different dates
order_date_check = df.copy()
order_date_check["Parsed Order Date"] = order_dates
order_date_check["Parsed Ship Date"] = ship_dates

print("\nUnique order IDs:", df["Order ID"].nunique())
print("Total rows:", len(df))

print("\nNumber of distinct ship dates per order:")
print(order_date_check.groupby("Order ID")["Parsed Ship Date"].nunique().value_counts().sort_index())
```

```python

# Check the distribution of lead times
print("Lead-time frequency:")
print(lead_time.value_counts().sort_index().head(10))

print("\nMost common lead times:")
print(lead_time.value_counts().head(10))

# Compare lead times with shipping mode
print("\nMedian lead time by shipping mode:")
print(date_analysis.groupby("Ship Mode")["Lead Time (days)"].median().sort_values())

# Inspect order IDs and dates together
print("\nOrder ID and date comparison:")
print(df[["Order ID", "Order Date", "Ship Date", "Ship Mode"]].head(15).to_string(index=False))
```

3. Calculated lead times range from 904 to 1,642 days.
4. Lead times are unusually long for the listed shipping modes, including Same Day.
5. Order IDs contain a year-like segment of 2021, which does not match the order-date range.
6. There are 8,549 unique orders across 10,194 rows, indicating that orders can contain multiple records.
7. Lead times show clusters around approximately 904, 1,273 and 1,639 days.
8. Median lead times are similar across all four shipping modes, suggesting that the date data may not represent actual shipping durations.
