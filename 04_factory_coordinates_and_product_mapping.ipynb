```python
import pandas as pd

df = pd.read_csv("/content/Nassau Candy Distributor.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

print("\nUnique products:")
display(sorted(df["Product Name"].unique()))
```

```python
factory_data = {
    "Factory": [
        "Lot's O' Nuts",
        "Wicked Choccy's",
        "Sugar Shack",
        "Secret Factory",
        "The Other Factory"
    ],

    "Latitude": [
        32.881893,
        32.076176,
        48.119140,
        41.446333,
        35.117500
    ],

    "Longitude": [
        -111.768036,
        -81.088371,
        -96.181150,
        -90.565487,
        -89.971107
    ]
}

factory_df = pd.DataFrame(factory_data)

print("FACTORY COORDINATES")
display(factory_df)
```

```python
product_factory_data = {
    "Product Name": [
        "Wonka Bar - Nutty Crunch Surprise",
        "Wonka Bar - Fudge Mallows",
        "Wonka Bar -Scrumdiddlyumptious",
        "Wonka Bar - Milk Chocolate",
        "Wonka Bar - Triple Dazzle Caramel",
        "Laffy Taffy",
        "SweeTARTS",
        "Nerds",
        "Fun Dip",
        "Fizzy Lifting Drinks",
        "Everlasting Gobstopper",
        "Hair Toffee",
        "Lickable Wallpaper",
        "Wonka Gum",
        "Kazookles"
    ],

    "Current Factory": [
        "Lot's O' Nuts",
        "Lot's O' Nuts",
        "Lot's O' Nuts",
        "Wicked Choccy's",
        "Wicked Choccy's",
        "Sugar Shack",
        "Sugar Shack",
        "Sugar Shack",
        "Sugar Shack",
        "Sugar Shack",
        "Secret Factory",
        "The Other Factory",
        "Secret Factory",
        "Secret Factory",
        "The Other Factory"
    ]
}

product_factory_df = pd.DataFrame(product_factory_data)

print("PRODUCT-TO-FACTORY MAPPING")
display(product_factory_df)
```

```python
print("MAPPING VALIDATION")
print("-------------------------")

print("Total products:", len(product_factory_df))

print(
    "Unique products:",
    product_factory_df["Product Name"].nunique()
)

print(
    "Missing values:",
    product_factory_df.isnull().sum().sum()
)

print(
    "Duplicate products:",
    product_factory_df["Product Name"].duplicated().sum()
)

print(
    "Invalid factory names:",
    len(
        set(product_factory_df["Current Factory"])
        - set(factory_df["Factory"])
    )
)
```

```python
combined_df = pd.merge(
    product_factory_df,
    factory_df,
    left_on="Current Factory",
    right_on="Factory",
    how="left"
)

combined_df = combined_df.drop(columns=["Factory"])

print("PRODUCT, FACTORY AND LOCATION DETAILS")
display(combined_df)
```

```python
print("COMBINED DATA VALIDATION")
print("------------------------------")

print("Total records:", len(combined_df))

print(
    "Missing values:",
    combined_df.isnull().sum().sum()
)

print(
    "Products without coordinates:",
    combined_df[["Latitude", "Longitude"]].isnull().any(axis=1).sum()
)

print(
    "Unique factories:",
    combined_df["Current Factory"].nunique()
)

print(
    "Unique products:",
    combined_df["Product Name"].nunique()
)
```

```python
factory_summary = (
    combined_df.groupby("Current Factory")["Product Name"]
    .count()
    .reset_index(name="Number of Products")
)

factory_summary = factory_summary.sort_values(
    by="Number of Products",
    ascending=False
)

print("FACTORY-WISE PRODUCT DISTRIBUTION")
print("---------------------------------")

display(factory_summary)
```

```python
combined_df.to_csv(
    "combined_product_factory_mapping.csv",
    index=False
)

print("Combined mapping saved successfully!")
```
